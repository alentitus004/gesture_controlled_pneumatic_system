```python
import pyfirmata

comport = 'COM8'
board = pyfirmata.Arduino(comport)

# LED pins
led_1 = board.get_pin('d:8:o')
led_2 = board.get_pin('d:9:o')
led_3 = board.get_pin('d:10:o')
led_4 = board.get_pin('d:11:o')
led_5 = board.get_pin('d:12:o')

# Relay control pin
relay = board.get_pin('d:7:o')


def led(fingerUp):
    count = sum(fingerUp)

    # LED indication
    led_1.write(1 if count >= 1 else 0)
    led_2.write(1 if count >= 2 else 0)
    led_3.write(1 if count >= 3 else 0)
    led_4.write(1 if count >= 4 else 0)
    led_5.write(1 if count >= 5 else 0)

    # Pneumatic cylinder control
    if count == 5:
        relay.write(1)  # Energize valve: extension command
    elif count == 0:
        relay.write(0)  # De-energize valve: spring-return command
```
